```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('student.csv')
df.head()
```

```python
plt.hist(df['StudyHours'])
```

```python
plt.scatter(df['StudyHours'],df['AssignmentScore'])
plt.xlabel('Study Hours')
plt.ylabel('Assignment Score')
plt.title('Study Hours vs Assignment Score')
plt.show()
```

```python
df.plot(
    kind='scatter',
    x='StudyHours',
    y='AssignmentScore'
)
```

```python
result_mean = df.groupby('Result')['StudyHours'].mean()
plt.bar(result_mean.index, result_mean.values)
plt.xticks([0,1])
plt.ylabel('Average Study Hours')
plt.title('Average Study Hours by Result')
plt.show()
```

```python
plt.scatter(df['Attendance'],df['AssignmentScore'])
plt.xlabel('Attendance')
plt.ylabel('AssignmentScore')
plt.title('Attendanve and Assignment Score')
plt.show()
```

```python
plt.hist(df['SleepHours'])
plt.xlabel('Sleep Hours')
plt.ylabel('Frequency')
plt.title('Sleep Hours Distribution')
plt.show()
```
