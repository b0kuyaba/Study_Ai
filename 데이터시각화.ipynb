```python
import matplotlib.pyplot as pit
x_data = [2,3,4,5]
y_data = [1,3,5,7]

pit.plot(x_data,y_data, label = 'data')
pit.ylim(0,8)
pit.title('Graph Title')
pit.xlabel('x_name')
pit.ylabel('y_name')
pit.legend()
pit.show()
```

```python
import pandas as pd
import matplotlib.pyplot as pit

data = pd.read_excel("dust_pm25.xlsx")

pit.figure(figsize=(9,3))

for year in range(2019,2022):
    y_data=data[year]
    pit.plot(y_data,marker = 's',label=year)
pit.xlabel('area')
pit.ylabel('micrometer')
pit.xticks(rotation=45)
pit.title('2019-2021 Dust Line Graph')
pit.legend()
pit.show()
```

```python
data2017 = data[2017]
pit.hist(data2017, bins=9, label='pm2.5')
pit.xlabel('pm2.5')
pit.ylabel('frequency')
pit.title('2017 Dust Histogram')
pit.legend()
pit.grid()

pit.show()
```
